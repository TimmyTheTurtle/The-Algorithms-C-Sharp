## 2. Fibonacci

**What it is:** Calculate the nth Fibonacci number: 0, 1, 1, 2, 3, 5, 8, 13, ...

**Why it matters for LLMs:**
- Classic recursion vs dynamic programming trade-off
- Demonstrates exponential complexity problems (naive recursion is SLOW)
- Foundation for understanding memoization and optimization
- Real-world optimization: same pattern appears in many LLM augmentation problems

**Algorithm walkthrough:**
- Definition: fib(0) = 0, fib(1) = 1, fib(n) = fib(n-1) + fib(n-2)
- Naive recursion: beautiful but slow (exponential time)
- Iterative: much faster (linear time)
- Coming later: memoization (cache results to avoid recalculation)

**Time complexity:** O(2^n) naive, O(n) iterative
**Space complexity:** O(n) recursion stack (naive), O(1) iterative

---

#### Recursive

```python
def fibonacci(n):
    """
    Calculate the nth Fibonacci number.
    
    Args:
        n: Non-negative integer
    
    Returns:
        The nth Fibonacci number
    """
    # Your code here
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)


# Test your implementation
print(fibonacci(0))   # Should display: 0
print(fibonacci(1))   # Should display: 1
print(fibonacci(2))   # Should display: 1
print(fibonacci(3))   # Should display: 2
print(fibonacci(4))   # Should display: 3
print(fibonacci(5))   # Should display: 5
print(fibonacci(10))  # Should display: 55
```

#### Iterative

```python
def fibonacci_iterative(n):
    if n == 0:
        return 0
    if n == 1:
        return 1

    # Keep track of two consecutive Fibonacci numbers:
    # previous is the number before current.
    previous, current = 0, 1

    # Start at the third Fibonacci number (index 2) and continue through n.
    for _ in range(2, n + 1):
        temp = current
        current = previous + current
        previous = temp

    # current now contains the nth Fibonacci number.
    return current

# Test your implementation
assert(fibonacci_iterative(10) == 55)

print(fibonacci_iterative(0))   # Should display: 0
print(fibonacci_iterative(1))   # Should display: 1
print(fibonacci_iterative(2))   # Should display: 1
print(fibonacci_iterative(5))   # Should display: 5
print(fibonacci_iterative(10))  # Should display: 55
```

## 3. Greatest Common Divisor (GCD)

**What it is:** Find the largest number that divides both a and b evenly.

**Why it matters for LLMs:**
- Validation and constraint satisfaction (do these numbers have certain properties?)
- Simplification (reducing fractions, scaling ratios)
- Number theory foundation (used in many algorithm design patterns)
- Classic algorithm: Euclidean algorithm (elegant and efficient)

**Algorithm walkthrough:**
- Euclidean algorithm: gcd(a, b) = gcd(b, a mod b) when b ≠ 0, and gcd(a, 0) = a
- Example: gcd(48, 18)
  - gcd(48, 18) = gcd(18, 48 mod 18) = gcd(18, 12)
  - gcd(18, 12) = gcd(12, 6)
  - gcd(12, 6) = gcd(6, 0) = 6

**Why Euclidean algorithm works:**
- Every common divisor of a and b is also a divisor of a mod b
- So the set of common divisors doesn't change

**Time complexity:** O(log(min(a, b)))
**Space complexity:** O(1) iterative, O(log(min(a, b))) recursive

---

```python
def gcd(a, b):
    """
    Calculate the Greatest Common Divisor of a and b using Euclidean algorithm.
    
    Args:
        a, b: Positive integers
    
    Returns:
        The GCD of a and b
    """
    # Your code here
    if b == 0:
        return a

    return gcd(b, a % b)


# Test your implementation
print(gcd(48, 18))    # Should display: 6
print(gcd(100, 50))   # Should display: 50
print(gcd(17, 19))    # Should display: 1 (coprime)
print(gcd(0, 5))      # Should display: 5
print(gcd(12, 12))    # Should display: 12
```
