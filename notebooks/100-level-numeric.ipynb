## 1. Factorial - Multiply all integers from 1 to n

```csharp
// Recursive implementation
decimal RecursiveFactorial(int n)
{
    if (n <= 1) return 1;
    return n * RecursiveFactorial(n - 1);
}

// Iterative implementation (more efficient)
decimal IterativeFactorial(int n)
{
    decimal result = 1;
    for (int i = 2; i <= n; i++)
        result *= i;
    return result;
}

// Test cases
Console.WriteLine("=== Factorial Examples ===");
for (int n = 0; n <= 12; n++)
{
    var factorial = IterativeFactorial(n);
    Console.WriteLine($"{n}! = {factorial}");
}
```

### Factorial with BigInteger (for huge numbers)

```csharp
using System.Numerics;

BigInteger BigFactorial(int n)
{
    BigInteger result = 1;
    for (int i = 2; i <= n; i++)
        result *= i;
    return result;
}

var result100 = BigFactorial(100);
Console.WriteLine($"100! has {result100.ToString().Length} digits");
Console.WriteLine($"100! = {result100}");
```

## 2. Fibonacci Sequence - Each number is sum of the two before it

```csharp
// Naive recursive (SLOW - exponential time)
int FibRecursive(int n)
{
    if (n <= 1) return n;
    return FibRecursive(n - 1) + FibRecursive(n - 2);
}

// Iterative (FAST - linear time)
int FibIterative(int n)
{
    if (n <= 1) return n;
    int prev = 0, curr = 1;
    for (int i = 2; i <= n; i++)
    {
        int next = prev + curr;
        prev = curr;
        curr = next;
    }
    return curr;
}

// Generate Fibonacci sequence
Console.WriteLine("First 15 Fibonacci Numbers:");
for (int n = 0; n < 15; n++)
{
    Console.WriteLine($"F({n}) = {FibIterative(n)}");
}
```

### Performance Comparison: Recursive vs Iterative

```csharp
var sw = System.Diagnostics.Stopwatch.StartNew();
var recursiveResult = FibRecursive(35);
sw.Stop();
Console.WriteLine($"Recursive F(35) = {recursiveResult}");
Console.WriteLine($"Time: {sw.ElapsedMilliseconds}ms");
Console.WriteLine($"Complexity: O(2^n) - exponential\n");

sw.Restart();
var iterativeResult = FibIterative(35);
sw.Stop();
Console.WriteLine($"Iterative F(35) = {iterativeResult}");
Console.WriteLine($"Time: {sw.ElapsedMilliseconds}ms");
Console.WriteLine($"Complexity: O(n) - linear");
```

## 3. Greatest Common Divisor (GCD) - Euclidean Algorithm

```csharp
int GCD(int a, int b)
{
    a = Math.Abs(a);
    b = Math.Abs(b);
    
    while (b != 0)
    {
        int temp = b;
        b = a % b;
        a = temp;
    }
    return a;
}

// Test cases
Console.WriteLine("=== GCD Examples ===");
var testCases = new[] { (48, 18), (100, 50), (17, 19), (1071, 462) };
foreach (var (a, b) in testCases)
{
    Console.WriteLine($"GCD({a}, {b}) = {GCD(a, b)}");
}
```

## 4. Prime Number Check

```csharp
bool IsPrime(int n)
{
    if (n < 2) return false;
    if (n == 2) return true;
    if (n % 2 == 0) return false;
    
    // Check odd divisors up to sqrt(n)
    for (int i = 3; i * i <= n; i += 2)
        if (n % i == 0) return false;
    
    return true;
}

// Find all primes up to 100
Console.WriteLine("Prime numbers from 1-100:");
var primes = Enumerable.Range(1, 100).Where(IsPrime).ToList();
Console.WriteLine(string.Join(", ", primes));
Console.WriteLine($"\nTotal: {primes.Count} primes");
```

## 5. Modular Exponentiation

Calculate (base^exp) mod n efficiently. Critical for cryptography.

```csharp
using System.Numerics;

BigInteger ModPow(BigInteger baseNum, BigInteger exponent, BigInteger modulus)
{
    BigInteger result = 1;
    baseNum = baseNum % modulus;
    
    while (exponent > 0)
    {
        if (exponent % 2 == 1)
            result = (result * baseNum) % modulus;
        
        exponent = exponent >> 1;  // Divide by 2
        baseNum = (baseNum * baseNum) % modulus;
    }
    
    return result;
}

// Examples
Console.WriteLine($"2^10 mod 1000 = {ModPow(2, 10, 1000)}");
Console.WriteLine($"3^100 mod 7 = {ModPow(3, 100, 7)}");
Console.WriteLine($"\nUsed in: cryptography, hash functions, modular arithmetic");
```
