## 1. Factorial

**What it is:**
Calculate n! (n factorial) = n × (n-1) × (n-2) × ... × 2 × 1

**Examples:**
- 5! = 5 × 4 × 3 × 2 × 1 = 120
- 0! = 1 (by definition)
- 1! = 1

**Why it matters for LLMs:**
- Understanding recursive algorithms (base case and recursive step)
- Combinatorics in decision-making (how many ways to arrange things?)

## 3. Greatest Common Divisor (GCD)

**What it is:**
Find the largest number that divides both input numbers evenly.

**Examples:**
- GCD(12, 8) = 4 (both 12 and 8 are divisible by 4)
- GCD(17, 5) = 1 (they share no common divisor except 1)
- GCD(100, 50) = 50

**Why it matters for LLMs:**
- Simplifying fractions/ratios
- Constraint satisfaction (finding compatible values)
- Number theory foundation
- Scheduling/timing problems (least common multiples)

**Algorithms:**
1. **Naive:** Check every number from 1 to min(a,b), keep largest divisor

- GCD(a, 0) = a
