## 3. Greatest Common Divisor (GCD)

**What it is:**
Find the largest number that divides both a and b evenly (with no remainder).

**Examples:**
- GCD(12, 8) = 4 (largest divisor of both)
- GCD(17, 5) = 1 (coprime: no common divisor except 1)
- GCD(100, 50) = 50

**Algorithm (Euclidean):**
1. Divide a by b, get remainder r
2. If r = 0, b is the GCD
3. If r ≠ 0, set a=b and b=r, repeat

**Why it matters for LLMs:**
- Simplifying fractions (a/b reduced form)
- Finding patterns in sequences
- Number theory applications in cryptography
