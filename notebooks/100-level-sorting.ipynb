## 1. Bubble Sort - Repeatedly step through and swap adjacent elements

```csharp
int[] BubbleSort(int[] arr)
{
    int[] result = (int[])arr.Clone();
    int n = result.Length;
    
    for (int i = 0; i < n - 1; i++)
    {
        for (int j = 0; j < n - i - 1; j++)
        {
            if (result[j] > result[j + 1])
            {
                // Swap
                int temp = result[j];
                result[j] = result[j + 1];
                result[j + 1] = temp;
            }
        }
    }
    
    return result;
}

// Test
int[] testArr = { 5, 2, 8, 1, 9 };
Console.WriteLine("=== Bubble Sort ===");
Console.WriteLine($"Original: [{string.Join(", ", testArr)}]");
var sorted = BubbleSort(testArr);
Console.WriteLine($"Sorted:   [{string.Join(", ", sorted)}]");
Console.WriteLine($"Time Complexity: O(n²)");
Console.WriteLine($"Space Complexity: O(1)");
```

## 2. Selection Sort - Find minimum and move it to sorted portion

```csharp
int[] SelectionSort(int[] arr)
{
    int[] result = (int[])arr.Clone();
    int n = result.Length;
    
    for (int i = 0; i < n - 1; i++)
    {
        // Find minimum in unsorted portion
        int minIndex = i;
        for (int j = i + 1; j < n; j++)
        {
            if (result[j] < result[minIndex])
                minIndex = j;
        }
        
        // Swap with current position
        int temp = result[i];
        result[i] = result[minIndex];
        result[minIndex] = temp;
    }
    
    return result;
}

// Test
int[] testArr2 = { 64, 25, 12, 22, 11 };
Console.WriteLine("=== Selection Sort ===");
Console.WriteLine($"Original: [{string.Join(", ", testArr2)}]");
var sorted2 = SelectionSort(testArr2);
Console.WriteLine($"Sorted:   [{string.Join(", ", sorted2)}]");
Console.WriteLine($"Time Complexity: O(n²) - always");
Console.WriteLine($"Advantage: Minimal swaps - good when write operations are expensive");
```

## 3. Insertion Sort - Build sorted array one item at a time

```csharp
int[] InsertionSort(int[] arr)
{
    int[] result = (int[])arr.Clone();
    
    for (int i = 1; i < result.Length; i++)
    {
        int key = result[i];
        int j = i - 1;
        
        // Shift elements greater than key to the right
        while (j >= 0 && result[j] > key)
        {
            result[j + 1] = result[j];
            j--;
        }
        
        // Insert key at correct position
        result[j + 1] = key;
    }
    
    return result;
}

// Test
int[] testArr3 = { 5, 2, 8, 1, 9 };
Console.WriteLine("=== Insertion Sort ===");
Console.WriteLine($"Original: [{string.Join(", ", testArr3)}]");
var sorted3 = InsertionSort(testArr3);
Console.WriteLine($"Sorted:   [{string.Join(", ", sorted3)}]");
Console.WriteLine($"Time Complexity: O(n²), but O(n) for nearly sorted data");
Console.WriteLine($"Advantage: Efficient for small arrays and nearly sorted data");
```

## 4. QuickSort - Divide and Conquer Approach

```csharp
int[] QuickSort(int[] arr, int low = 0, int high = -1)
{
    if (high == -1) high = arr.Length - 1;
    int[] result = (int[])arr.Clone();
    
    void QuickSortHelper(int l, int h)
    {
        if (l < h)
        {
            int pi = Partition(l, h);
            QuickSortHelper(l, pi - 1);
            QuickSortHelper(pi + 1, h);
        }
    }
    
    int Partition(int l, int h)
    {
        int pivot = result[h];
        int i = l - 1;
        
        for (int j = l; j < h; j++)
        {
            if (result[j] < pivot)
            {
                i++;
                int temp = result[i];
                result[i] = result[j];
                result[j] = temp;
            }
        }
        
        int temp2 = result[i + 1];
        result[i + 1] = result[h];
        result[h] = temp2;
        
        return i + 1;
    }
    
    QuickSortHelper(low, high);
    return result;
}

// Test
int[] testArr4 = { 5, 2, 8, 1, 9, 3 };
Console.WriteLine("=== QuickSort ===");
Console.WriteLine($"Original: [{string.Join(", ", testArr4)}]");
var sorted4 = QuickSort(testArr4);
Console.WriteLine($"Sorted:   [{string.Join(", ", sorted4)}]");
Console.WriteLine($"Time Complexity: O(n log n) average, O(n²) worst case");
Console.WriteLine($"Space Complexity: O(log n) - recursion stack");
```

## 5. MergeSort - Guaranteed O(n log n)

```csharp
int[] MergeSort(int[] arr)
{
    if (arr.Length <= 1) return arr;
    
    int mid = arr.Length / 2;
    int[] left = MergeSort(arr.Take(mid).ToArray());
    int[] right = MergeSort(arr.Skip(mid).ToArray());
    
    return Merge(left, right);
}

int[] Merge(int[] left, int[] right)
{
    int[] result = new int[left.Length + right.Length];
    int i = 0, j = 0, k = 0;
    
    while (i < left.Length && j < right.Length)
    {
        if (left[i] <= right[j])
            result[k++] = left[i++];
        else
            result[k++] = right[j++];
    }
    
    while (i < left.Length) result[k++] = left[i++];
    while (j < right.Length) result[k++] = right[j++];
    
    return result;
}

// Test
int[] testArr5 = { 5, 2, 8, 1, 9 };
Console.WriteLine("=== MergeSort ===");
Console.WriteLine($"Original: [{string.Join(", ", testArr5)}]");
var sorted5 = MergeSort(testArr5);
Console.WriteLine($"Sorted:   [{string.Join(", ", sorted5)}]");
Console.WriteLine($"Time Complexity: O(n log n) - always guaranteed");
Console.WriteLine($"Space Complexity: O(n) - requires extra array");
Console.WriteLine($"Advantage: Stable and predictable, good for linked lists");
```

## Performance Comparison

```csharp
// Generate test data
var random = new Random(42);
int[] RandomArray(int size) => Enumerable.Range(0, size).Select(_ => random.Next(1000)).ToArray();

var testData = RandomArray(100);

Console.WriteLine("=== Sorting Performance (100 elements) ===");

var sw1 = System.Diagnostics.Stopwatch.StartNew();
BubbleSort(testData);
sw1.Stop();
Console.WriteLine($"Bubble Sort:   {sw1.ElapsedTicks,8} ticks");

var sw2 = System.Diagnostics.Stopwatch.StartNew();
SelectionSort(testData);
sw2.Stop();
Console.WriteLine($"Selection Sort: {sw2.ElapsedTicks,8} ticks");

var sw3 = System.Diagnostics.Stopwatch.StartNew();
InsertionSort(testData);
sw3.Stop();
Console.WriteLine($"Insertion Sort: {sw3.ElapsedTicks,8} ticks");

var sw4 = System.Diagnostics.Stopwatch.StartNew();
QuickSort(testData);
sw4.Stop();
Console.WriteLine($"QuickSort:     {sw4.ElapsedTicks,8} ticks");

var sw5 = System.Diagnostics.Stopwatch.StartNew();
MergeSort(testData);
sw5.Stop();
Console.WriteLine($"MergeSort:     {sw5.ElapsedTicks,8} ticks");
```

## 💡 Learning Tips

✓ Elementary sorts (Bubble, Selection, Insertion) = O(n²) - OK for small data  
✓ Advanced sorts (Quick, Merge) = O(n log n) - Necessary for large data  
✓ Stability = preserves relative order of equal elements  
✓ Space = in-place O(1) vs requiring extra arrays  
✓ C#'s Sort() uses Introsort (Quick + Heap + Insertion hybrid)
