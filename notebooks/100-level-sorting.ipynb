## 1. Bubble Sort

**What it is:** Repeatedly step through list, compare adjacent elements, swap if needed.

**Why it matters for LLMs:**
- Foundation for understanding comparison-based sorting
- Demonstrates why naive approaches are inefficient
- Good for nearly-sorted data (adaptive: O(n) if already sorted)
- Simple enough to verify correctness

**Algorithm walkthrough:**
- Outer loop: n passes through the list
- Inner loop: compare each pair of adjacent elements
- If current > next, swap them
- After each pass, largest unsorted element "bubbles" to end
- Optimization: reduce inner loop range each pass (already-sorted portion grows)

**Time complexity:** O(n²) average/worst, O(n) best (already sorted)
**Space complexity:** O(1)
**Stable:** Yes (equal elements maintain order)

---

```python
def bubble_sort(arr):
    """
    Sort array using bubble sort algorithm.
    
    Args:
        arr: List of comparable elements
    
    Returns:
        Sorted list
    """
    # Your code here
    res = arr.copy()
    swap = True
    while swap:
        swap = False
        for i in range(0, len(res) - 1):
            if res[i]>res[i + 1]:
                temp = res[i]
                res[i] = res[i + 1]
                res[i + 1] = temp
                swap = True

    return res


# Test your implementation
print(bubble_sort([64, 34, 25, 12, 22, 11, 90]))  # Should display: [11, 12, 22, 25, 34, 64, 90]
print(bubble_sort([5, 2, 8, 1, 9]))               # Should display: [1, 2, 5, 8, 9]
print(bubble_sort([1]))                           # Should display: [1]
print(bubble_sort([]))                            # Should display: []
```

## 2. Selection Sort

**What it is:** Divide list into sorted and unsorted regions. Repeatedly find minimum in unsorted region, move to sorted region.

**Why it matters for LLMs:**
- Different sorting strategy (find minimum vs pairwise comparisons)
- Minimizes number of writes (good for expensive write operations)
- Demonstrates that time complexity alone doesn't tell full story
- Always O(n²) regardless of input (predictable, no best-case advantage)

**Algorithm walkthrough:**
- Outer loop: iterate through each position from start to end
- Inner loop: find minimum element in remaining unsorted portion
- Swap minimum with current position
- Move boundary between sorted/unsorted one position right

**Time complexity:** O(n²) all cases
**Space complexity:** O(1)
**Stable:** No (swapping can change order of equal elements)

---

```python
def selection_sort(arr):
    """
    Sort array using selection sort algorithm.
    
    Args:
        arr: List of comparable elements
    
    Returns:
        Sorted list
    """
    # Your code here
    res = arr.copy()

    # You also stop at len(res) - 1 
    # because once every earlier position is sorted, 
    # the final element is automatically in the correct position.
    for i in range(len(res) - 1):
        #unsorted = res[i:].copy()
        min_index = min(range(i, len(res)), key=res.__getitem__)
        if min_index > i:
            temp = res[min_index]
            res[min_index] = res[i]
            res[i] = temp

    return res


# Test your implementation
print(selection_sort([64, 34, 25, 12, 22, 11, 90]))  # Should display: [11, 12, 22, 25, 34, 64, 90]
print(selection_sort([5, 2, 8, 1, 9]))               # Should display: [1, 2, 5, 8, 9]
print(selection_sort([1]))                           # Should display: [1]
print(selection_sort([]))                            # Should display: []
```

**Why it matters for LLMs:**

- Stable sort (preserves order of equal elements)

**Algorithm walkthrough:**
- Start with first element as sorted portion
- For each new element: find correct position in sorted portion, shift elements, insert
- After processing all elements, entire array is sorted

## 4. Quick Sort

**What it is:** Divide-and-conquer: pick pivot, partition around it, recursively sort partitions.

**Why it matters for LLMs:**
- Powerful divide-and-conquer pattern (used throughout computer science)
- Fast average case (O(n log n)) with low memory overhead
- Most real-world sorting uses quicksort variants (Python's built-in sort uses hybrid)
- Understanding pivot selection and partitioning strategy

**Algorithm walkthrough:**
- Choose pivot element (various strategies: first, last, random, median-of-three)
- Partition: rearrange so elements < pivot are left, elements > pivot are right
- Recursively sort left and right partitions
- Base case: arrays of 1 element are sorted

**Time complexity:** O(n log n) average, O(n²) worst (bad pivot choices), O(n log n) best
**Space complexity:** O(log n) recursion stack

## 5. Merge Sort

**What it is:** Divide-and-conquer: split array in half, recursively sort each half, merge sorted halves.

**Why it matters for LLMs:**
- Guaranteed O(n log n) performance (no worst-case surprises)
- Stable sort (equal elements maintain order)
- Foundation for external sorting (sorting data too large for memory)
- Demonstrates the space-time trade-off (uses more memory but guarantees performance)

**Algorithm walkthrough:**
- Base case: arrays of 1 element are sorted
- Divide: split array into two halves
- Conquer: recursively sort each half
- Combine (merge): merge two sorted halves into single sorted array

**Merge operation:**
- Compare elements from left and right halves
- Take smaller one, add to result, advance that pointer
- When one half exhausted, append remaining elements from other half

**Time complexity:** O(n log n) all cases

## Performance Comparison Cell

Once you've implemented all sorting algorithms, run this cell to compare their performance on different data sizes.

```python
import time
import random

def time_sort(sort_func, data, name):
    """Time a sorting function and return elapsed time."""
    start = time.time()
    result = sort_func(data.copy())
    elapsed = time.time() - start
    return elapsed


# Generate test data
sizes = [100, 500, 1000]
algorithms = [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort),
    ("Quick Sort", quick_sort),
    ("Merge Sort", merge_sort),
]

for size in sizes:
    data = [random.randint(1, 1000) for _ in range(size)]
    print(f"\n--- Array size: {size} ---")
    for name, func in algorithms:
        elapsed = time_sort(func, data, name)
        print(f"{name:20} {elapsed:.6f} seconds")
```
