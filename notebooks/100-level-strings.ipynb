## 2. Palindrome Check

**What it is:**
Check if a string reads the same forwards and backwards. A palindrome is symmetric—it has mirror symmetry in text.

**The algorithm:**
1. Clean the string (remove spaces, punctuation, make lowercase)
2. Compare the first character with the last
3. Compare the second with the second-to-last
4. Keep moving inward until you meet in the middle
5. If all match, it's a palindrome; if any don't match, it's not

- Data quality checks
- Pattern recognition in LLM responses

## 5. Anagram Detection

**What it is:**
Check if two strings contain the same letters, just rearranged. "listen" and "silent" are anagrams—same letters, different order.

**The algorithm:**
1. Clean both strings (lowercase, remove spaces/punctuation)
2. Check if they have the same length (if not, can't be anagrams)
3. Count character frequency in string 1
4. Count character frequency in string 2
5. If the frequency maps are identical, they're anagrams

**Alternative approach:**
1. Sort both strings
2. If sorted versions are equal, they're anagrams
