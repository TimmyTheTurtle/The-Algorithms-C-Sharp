## 1. Reverse String

**What it is:** Return a string in reverse order.

**Why it matters for LLMs:**
- Palindrome detection (validation tasks)
- String symmetry checks (pattern recognition)
- Foundation for more complex reversals (substring reversal, word reversal)

**Algorithm walkthrough:**
- Approach 1: Convert to list, reverse, rejoin
- Approach 2: Iterate from end to start, build new string
- Approach 3: Use Python slicing shortcut (for learning, understand the manual approach first)

**Time complexity:** O(n) where n is string length
**Space complexity:** O(n) for the reversed string

---

```python
def reverse_string(s):
    """
    Reverse a string.
    
    Args:
        s: Input string
    
    Returns:
        The reversed string
    """
    # Your code here
    res = []
    right_index = len(s) - 1
    while right_index >= 0:
        res.append(s[right_index])
        right_index -= 1
    return "".join(res)

def reverse_string_vowels(s):
    """
    Reverse only the vowels in a string, keeping consonants in place.
    
    Args:
        s: Input string
    
    Returns:
        The string with vowels reversed
    """
    vowels = set('aeiouAEIOU')
    s_list = list(s)
    left, right = 0, len(s_list) - 1
    
    while left < right:
        # Find next vowel from left
        while left < right and s_list[left] not in vowels:
            left += 1
        # Find next vowel from right
        while left < right and s_list[right] not in vowels:
            right -= 1
        # Swap vowels
        s_list[left], s_list[right] = s_list[right], s_list[left]
        left += 1
        right -= 1
    
    return ''.join(s_list)


# Test your implementation
print(reverse_string_vowels("hello"))      # Should display: olleh
print(reverse_string_vowels("Python"))     # Should display: nohtyP
print(reverse_string_vowels(""))           # Should display: (empty)
print(reverse_string_vowels("akejfiworu")) # Should display: a
print(reverse_string_vowels("12345"))      # Should display: 54321
print(reverse_string_vowels("racecar"))    # Should display: racecar (palindrome)
print(reverse_string_vowels("A"))          # Should display: A
print(reverse_string_vowels("Hello, World!"))  # Should display: !dlroW ,olleH
print(reverse_string_vowels("   "))        # Should display:     (spaces reversed)
print(reverse_string_vowels("!@#$%"))      # Should display: %$#@!
```

## 2. Is Palindrome

**What it is:** Check if a string reads the same forwards and backwards.

**Why it matters for LLMs:**
- Validation and constraint checking (does this phrase meet symmetry requirements?)
- Pattern recognition in generated text
- Quality control for structured outputs

**Algorithm walkthrough:**
- Compare character at position i with character at position n-1-i
- Two-pointer approach: start from both ends, move towards center
- Stop when pointers meet or mismatch found

**Edge cases:**
- Empty string (is it a palindrome? Yes, by definition)
- Single character (palindrome)
- Case sensitivity ("Racecar" vs "racecar")
- Spaces and punctuation (ignore or consider?)

**Time complexity:** O(n)
**Space complexity:** O(1) if checking in place, O(n) if creating reversed copy

---

```python
def is_palindrome(s):
    """
    Check if a string is a palindrome (ignoring spaces, case-insensitive).
    
    Args:
        s: Input string
    
    Returns:
        True if palindrome, False otherwise
    """
    # Your code here
    s = s.replace(" ", "").lower()
    left = 0
    right = len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True


# Test your implementation
print(is_palindrome("racecar"))      # Should display: True
print(is_palindrome("hello"))        # Should display: False
print(is_palindrome("A man a plan a canal Panama"))  # Should display: True (ignoring spaces)
print(is_palindrome(""))             # Should display: True
print(is_palindrome("a"))            # Should display: True
```

## 3. Count Characters

**What it is:** Count occurrences of each character in a string.

**Why it matters for LLMs:**
- Frequency analysis (tokenization, language properties)
- Character distribution validation
- Building frequency maps for encoding/compression
- Detecting duplicate characters (common prompt validation task)

**Algorithm walkthrough:**
- Use a dictionary (hash map) to store character → count pairs
- Iterate through string, increment count for each character
- Return the frequency map

**Data structure:**
- Dictionary/Hash Map: O(1) average lookup and insertion

**Time complexity:** O(n)
**Space complexity:** O(k) where k is the number of unique characters (max 26 for lowercase letters)

---

```python
def count_characters(s):
    """
    Count the frequency of each character in a string.
    
    Args:
        s: Input string
    
    Returns:
        Dictionary with character frequencies
    """
    # Your code here

    counts = {}

    for c in s:
        counts[c] = counts.get(c, 0) + 1

    return counts


# Test your implementation
result = count_characters("hello")
print(result)  # Should display: {'h': 1, 'e': 1, 'l': 2, 'o': 1}

result = count_characters("mississippi")
print(result)  # Should display: {'m': 1, 'i': 4, 's': 4, 'p': 2}

result = count_characters("")
print(result)  # Should display: {}
```

**Challenge Problem:**
- Find the character with the highest frequency
- What happens if there's a tie?
- Can you do it in a single pass (without iterating through the frequency map after counting)?

```python
def highest_frequency(s):
    counts = {}

    for c in s:
        counts[c] = counts.get(c, 0) + 1
        # sorted_counts` stores the `(character, count)` pairs sorted by count from highest to lowest.
        # - `counts.items()` provides pairs such as `('a', 4)`.
        # - `key=lambda item: item[1]` sorts using the count.
        # - `reverse=True` puts the largest count first.
    sorted_counts = sorted(
        counts.items(),
        key=lambda item: item[1],
        reverse=True
    )

    return sorted_counts[0] if sorted_counts else None


print(highest_frequency("gabba gabba hay!!!!!"))

print(highest_frequency(""))
```

```python
def highest_frequency_single_pass(s):
    counts = {}
    highest_c = ''
    highest_count = 0
    for c in s:
        counts[c] = counts.get(c, 0) + 1
        if counts[c] >= highest_count:
            highest_count = counts[c]
            highest_c = c

    return highest_c, highest_count

print(highest_frequency_single_pass("!gabba gabba hay!!!!"))

print(highest_frequency_single_pass(""))
```

## 4. Find Substring

**What it is:** Find the starting index of a substring within a string, or return -1 if not found.

**Why it matters for LLMs:**
- Text parsing and extraction
- Prompt template matching (where does the variable start?)
- String validation (does response contain required phrase?)
- Foundation for more complex pattern matching

**Algorithm walkthrough:**
- Naive approach: check every position in the string
- For each position i, check if substring matches starting at i
- If match found, return i; if we reach end, return -1

**Optimization hints:**
- Early termination: if we're at position i, we can't fit substring if i + len(substring) > len(string)
- More advanced: KMP algorithm (outside 100-level scope)

**Time complexity:** O(n*m) naive approach, where n is string length and m is substring length
**Space complexity:** O(1)

---

```python
def find_substring(text, substring):
    """
    Find the index of the first occurrence of substring in text.
    
    Args:
        text: String to search in
        substring: String to search for
    
    Returns:
        Index of first occurrence, or -1 if not found
    """
    for i in range(len(text) - len(substring) + 1):
        if text[i:i + len(substring)] == substring:
            return i

    return -1


# Test your implementation
print(find_substring("hello world", "world"))  # Should display: 6
print(find_substring("hello world", "hello"))  # Should display: 0
print(find_substring("hello world", "xyz"))    # Should display: -1
print(find_substring("", "a"))                 # Should display: -1
print(find_substring("a", ""))                 # Should display: 0 (empty substring at start)
```

## 5. Are Anagrams

**What it is:** Check if two strings are anagrams (same characters, different order).

**Why it matters for LLMs:**
- Similarity detection (semantically different but structurally identical)
- Data validation (checking if input rearrangement matches a pattern)
- Word game constraints in constrained generation
- Building blocks for more complex rearrangement problems

**Algorithm walkthrough:**
- Approach 1: Sort both strings and compare (simpler, slightly less efficient)
- Approach 2: Count character frequencies in both strings, compare dictionaries
- Both approaches handle case sensitivity and special characters

**Considerations:**
- Case sensitivity: is "Listen" an anagram of "Silent"?
- Spaces and punctuation: should they be ignored?

**Time complexity:** O(n log n) for sorting approach, O(n) for frequency approach
**Space complexity:** O(n) for both

---

```python
def are_anagrams(s1, s2):
    """
    Check if two strings are anagrams (case-insensitive, ignore spaces).
    
    Args:
        s1: First string
        s2: Second string
    
    Returns:
        True if anagrams, False otherwise
    """
    # Your code here
    string_1_counts = {}
    string_2_counts = {}

    for char1 in s1:
        string_1_counts[char1] = string_1_counts.get(char1, 0) + 1

    for char2 in s2:
        string_2_counts[char2] = string_2_counts.get(char2, 0) + 1;

    return string_1_counts == string_2_counts


# Test your implementation
print(are_anagrams("listen", "silent"))       # Should display: True
print(are_anagrams("hello", "world"))         # Should display: False
print(are_anagrams("abc", "cab"))             # Should display: True
print(are_anagrams("", ""))                   # Should display: True
print(are_anagrams("a", "b"))                 # Should display: False
```

## 6. Case Conversion

**What it is:** Convert strings between uppercase and lowercase.

**Why it matters for LLMs:**
- Normalization of input (standardizing text before processing)
- Output formatting (adhering to specific casing requirements)
- Case-insensitive comparisons ("Yes" vs "yes" should be equivalent)
- Text preprocessing step in many NLP tasks

**Algorithm walkthrough:**
- Character-by-character conversion using ASCII values
- Uppercase: A-Z are ASCII 65-90, lowercase: a-z are ASCII 97-122`
- Difference between uppercase and lowercase: 32
- Python has built-in .upper() and .lower() methods (understand the concept first, then use built-ins)

**Time complexity:** O(n)
**Space complexity:** O(n) for the new string

---

```python
case_span = 32

def convert_to_upper(s):
    """
    Convert string to uppercase.
    
    Args:
        s: Input string
    
    Returns:
        String in uppercase
    """
    # Your code here
    result = ""
    for char in s:
        if 'a' <= char <= 'z':
            result += chr(ord(char) - case_span)
        else:
            result += char
        
    return result


def convert_to_lower(s):
    """
    Convert string to lowercase.
    
    Args:
        s: Input string
    
    Returns:
        String in lowercase
    """
    # Your code here
    result = ""

    for char in s:
        if 'A' <= char <= 'Z':
            result += chr(ord(char) + case_span)
        else:
            result += char
    
    return result


# Test your implementation
print(convert_to_upper("Hello World"))  # Should display: HELLO WORLD
print(convert_to_lower("Hello World"))  # Should display: hello world
print(convert_to_upper("abc123"))       # Should display: ABC123
print(convert_to_lower("ABC123"))       # Should display: abc123
```

## 7. Count Vowels and Consonants

**What it is:** Count how many vowels and consonants are in a string.

**Why it matters for LLMs:**
- Text analysis and linguistic properties
- Readability scoring (vowel-to-consonant ratio affects pronounceability)
- Phonetic preprocessing for speech synthesis
- Building blocks for more complex linguistic analysis

**Definitions:**
- Vowels: A, E, I, O, U (and lowercase a, e, i, o, u)
- Consonants: All other letters
- Non-letters: Ignore spaces, numbers, punctuation

**Algorithm walkthrough:**
- Iterate through each character
- Check if character is a vowel → increment vowel count
- Check if character is a letter and not a vowel → increment consonant count
- Ignore non-letter characters

**Time complexity:** O(n)
**Space complexity:** O(1)

---

```python
def count_vowels_and_consonants(s):
    """
    Count vowels and consonants in a string.
    
    Args:
        s: Input string
    
    Returns:
        Tuple (vowel_count, consonant_count)
    """
    # Your code here


    
    # Uppercase: A-Z are ASCII 65-90, lowercase: a-z are ASCII 97-122
    upperCase_start = 65
    upperCase_end = 90
    lowerCase_start = 97
    lowerCase_end = 122
    vowel_count = 0
    cons_count = 0
    vowel_ords = [ord('a'),ord('A'),ord('e'),ord('E'),ord('i'),ord('I'),ord('o'),ord('O'),ord('u'),ord('U')]

    for char in s:
        if (ord(char) >=65 and ord(char) <= 90) or (ord(char) >= 97 and ord(char) <= 122):
            if ord(char) in vowel_ords:
                vowel_count += 1
            else:
                cons_count += 1

    return (vowel_count, cons_count)


# Test your implementation
vowels, consonants = count_vowels_and_consonants("Hello World")
print(f"Vowels: {vowels}, Consonants: {consonants}")  # Should display: Vowels: 3, Consonants: 7

vowels, consonants = count_vowels_and_consonants("aeiou")
print(f"Vowels: {vowels}, Consonants: {consonants}")  # Should display: Vowels: 5, Consonants: 0

vowels, consonants = count_vowels_and_consonants("bcdfg")
print(f"Vowels: {vowels}, Consonants: {consonants}")  # Should display: Vowels: 0, Consonants: 5

vowels, consonants = count_vowels_and_consonants("123 !@#")
print(f"Vowels: {vowels}, Consonants: {consonants}")  # Should display: Vowels: 0, Consonants: 0
```
