## 1. String Reversal - Read a string backwards

```csharp
// Method 1: Using LINQ
string ReverseBuiltin(string s) => new string(s.Reverse().ToArray());

// Method 2: Manual two-pointer swap
string ReverseManual(string s)
{
    char[] chars = s.ToCharArray();
    int left = 0, right = chars.Length - 1;
    
    while (left < right)
    {
        char temp = chars[left];
        chars[left] = chars[right];
        chars[right] = temp;
        left++;
        right--;
    }
    
    return new string(chars);
}

// Test cases
Console.WriteLine("=== String Reversal ===");
var testStrings = new[] { "hello", "racecar", "LINQPad", "12345", "" };

foreach (var s in testStrings)
{
    var reversed = ReverseManual(s);
    var isPalindrome = s == reversed ? "✓ Palindrome" : "";
    Console.WriteLine($"{s,15} → {reversed,15} {isPalindrome}");
}
```

## 2. Palindrome Check - Words that read the same forwards and backwards

```csharp
bool IsPalindrome(string s)
{
    // Remove non-alphanumeric and convert to lowercase
    string cleaned = new string(s.Where(char.IsLetterOrDigit).ToArray()).ToLower();
    
    // Compare with reverse
    return cleaned == new string(cleaned.Reverse().ToArray());
}

// Test cases
Console.WriteLine("=== Palindrome Check ===");
var palindromeTests = new[] 
{ 
    "racecar",
    "hello",
    "A man, a plan, a canal: Panama",
    "Was it a car or a cat I saw?",
    "12321",
    "hello world"
};

foreach (var text in palindromeTests)
{
    var result = IsPalindrome(text) ? "✓ YES" : "✗ NO";
    Console.WriteLine($"{result} - {text}");
}
```

## 3. Character Frequency Count - How many times each character appears

```csharp
Dictionary<char, int> CharacterFrequency(string s)
{
    var frequency = new Dictionary<char, int>();
    
    foreach (char c in s.ToLower())
    {
        if (char.IsLetterOrDigit(c))
        {
            if (frequency.ContainsKey(c))
                frequency[c]++;
            else
                frequency[c] = 1;
        }
    }
    
    return frequency.OrderByDescending(x => x.Value).ToDictionary(x => x.Key, x => x.Value);
}

// Test case
var text = "hello world";
var freq = CharacterFrequency(text);

Console.WriteLine($"=== Character Frequency in \"{text}\" ===");
foreach (var kvp in freq)
{
    var percentage = (kvp.Value * 100.0 / text.Length);
    Console.WriteLine($"'{kvp.Key}': {kvp.Value} times ({percentage:F1}%)");
}
```

## 4. Anagrams - Same letters rearranged

```csharp
bool AreAnagrams(string s1, string s2)
{
    // Sort characters and compare
    var sorted1 = new string(s1.ToLower().Where(char.IsLetter).OrderBy(c => c).ToArray());
    var sorted2 = new string(s2.ToLower().Where(char.IsLetter).OrderBy(c => c).ToArray());
    
    return sorted1 == sorted2;
}

// Test cases
Console.WriteLine("=== Anagram Check ===");
var anagramTests = new[] 
{ 
    ("listen", "silent"),
    ("hello", "world"),
    ("abc", "bca"),
    ("The Eyes", "They See"),
    ("dormitory", "dirty room")
};

foreach (var (s1, s2) in anagramTests)
{
    var result = AreAnagrams(s1, s2) ? "✓ YES" : "✗ NO";
    Console.WriteLine($"{result} - {s1,20} and {s2}");
}
```

## 5. Substring Search - Find text within text

```csharp
// Manual implementation (for learning)
int ManualIndexOf(string haystack, string needle)
{
    if (needle.Length > haystack.Length) return -1;
    
    for (int i = 0; i <= haystack.Length - needle.Length; i++)
    {
        bool match = true;
        for (int j = 0; j < needle.Length; j++)
        {
            if (char.ToLower(haystack[i + j]) != char.ToLower(needle[j]))
            {
                match = false;
                break;
            }
        }
        if (match) return i;
    }
    
    return -1;
}

// Test cases
Console.WriteLine("=== Substring Search ===");
var searchTests = new[] 
{ 
    ("The quick brown fox", "brown"),
    ("C# Notebooks are awesome", "PAD"),
    ("Learning C#", "xyz"),
    ("AAAA", "AA"),
};

foreach (var (text, search) in searchTests)
{
    var index = ManualIndexOf(text, search);
    var result = index >= 0 ? $"✓ Found at index {index}" : "✗ Not found";
    Console.WriteLine($"{result,30} - Search \"{search}\" in \"{text}\"");
}
```

## 6. Vowel & Consonant Count

```csharp
(int vowels, int consonants) CountVowelsConsonants(string s)
{
    int vowelCount = 0, consonantCount = 0;
    var vowelSet = "aeiouAEIOU";
    
    foreach (char c in s)
    {
        if (char.IsLetter(c))
        {
            if (vowelSet.Contains(c))
                vowelCount++;
            else
                consonantCount++;
        }
    }
    
    return (vowelCount, consonantCount);
}

// Test
string testPhrase = "Learning algorithms is fun!";
var (v, c) = CountVowelsConsonants(testPhrase);

Console.WriteLine($"=== Vowel & Consonant Count ===");
Console.WriteLine($"Text: {testPhrase}");
Console.WriteLine($"Vowels: {v}");
Console.WriteLine($"Consonants: {c}");
Console.WriteLine($"Total letters: {v + c}");
Console.WriteLine($"Vowel ratio: {(v * 100.0 / (v + c)):F1}%");
```

## 7. Case Conversion

```csharp
string ToggleCase(string s) => new string(s.Select(c => char.IsUpper(c) ? char.ToLower(c) : char.ToUpper(c)).ToArray());
string ToCamelCase(string s) => char.ToLower(s[0]) + s.Substring(1);
string ToPascalCase(string s) => char.ToUpper(s[0]) + s.Substring(1);

// Examples
string example = "helloWorld";

Console.WriteLine("=== Case Conversion ===");
Console.WriteLine($"Original:      {example}");
Console.WriteLine($"ToUpper:       {example.ToUpper()}");
Console.WriteLine($"ToLower:       {example.ToLower()}");
Console.WriteLine($"ToggleCase:    {ToggleCase(example)}");
Console.WriteLine($"ToCamelCase:   {ToCamelCase("HELLO_WORLD")}");
Console.WriteLine($"ToPascalCase:  {ToPascalCase("hello_world")}");
```
