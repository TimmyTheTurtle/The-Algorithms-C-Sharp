## Encoding, encryption, and ciphers

An **encoding** changes a representation so systems can process it. A **cipher** transforms plaintext into ciphertext using a key. The encoders in this notebook are historical substitution ciphers: each letter is replaced according to a predictable rule.

Use modern, reviewed cryptographic libraries for real security. We study these algorithms to practice indexing, modular arithmetic, and reversible transformations.

## 1. Caesar cipher

A Caesar cipher shifts every letter by the same number. With a shift of 3, `A` becomes `D` and `Z` wraps around to `C`. Non-letter characters stay unchanged.

**Time complexity:** O(n), where `n` is the text length.

**Space complexity:** O(n) for the output.

```python
def caesar_encode(text, shift):
    """Shift alphabetic characters while preserving case and punctuation."""
    encoded = []
    for character in text:
        if not character.isalpha() or not character.isascii():
            encoded.append(character)
            continue

        first_letter = ord("A") if character.isupper() else ord("a")
        offset = ord(character) - first_letter
        encoded.append(chr(first_letter + (offset + shift) % 26))

    return "".join(encoded)


def caesar_decode(text, shift):
    """Reverse a Caesar shift by shifting in the opposite direction."""
    return caesar_encode(text, -shift)


plaintext = "Attack at Dawn!"
ciphertext = caesar_encode(plaintext, 3)
print("Plaintext: ", plaintext)
print("Ciphertext:", ciphertext)
print("Decoded:   ", caesar_decode(ciphertext, 3))

assert ciphertext == "Dwwdfn dw Gdzq!"
assert caesar_decode(ciphertext, 3) == plaintext
assert caesar_encode("Zebra", 2) == "Bgdtc"
```

## 2. Vigenere cipher

The Vigenere cipher repeats a word as its key. Each key letter supplies a different Caesar shift: `A` means shift 0, `B` means shift 1, and so on. A repeating key is more varied than a single shift, but it is still not secure by modern standards.

This version advances the key only when it processes an ASCII letter, so spaces and punctuation remain in their original positions.

**Time complexity:** O(n).

**Space complexity:** O(n).

```python
def vigenere_encode(text, key):
    """Encode ASCII letters with repeating alphabetic key shifts."""
    if not key or not key.isalpha() or not key.isascii():
        raise ValueError("key must be a non-empty ASCII alphabetic string")

    encoded = []
    key_index = 0
    for character in text:
        if not character.isalpha() or not character.isascii():
            encoded.append(character)
            continue

        key_character = key[key_index % len(key)].upper()
        shift = ord(key_character) - ord("A")
        encoded.append(caesar_encode(character, shift))
        key_index += 1

    return "".join(encoded)


def vigenere_decode(text, key):
    """Decode text by applying each repeating key shift in reverse."""
    if not key or not key.isalpha() or not key.isascii():
        raise ValueError("key must be a non-empty ASCII alphabetic string")

    decoded = []
    key_index = 0
    for character in text:
        if not character.isalpha() or not character.isascii():
            decoded.append(character)
            continue

        key_character = key[key_index % len(key)].upper()
        shift = ord(key_character) - ord("A")
        decoded.append(caesar_decode(character, shift))
        key_index += 1

    return "".join(decoded)


message = "Meet me at the park at eleven am."
key = "LEMON"
encoded_message = vigenere_encode(message, key)
print("Plaintext: ", message)
print("Ciphertext:", encoded_message)
print("Decoded:   ", vigenere_decode(encoded_message, key))

assert vigenere_decode(encoded_message, key) == message
assert vigenere_encode("AttackAtDawn", "LEMON") == "LxfopvEfRnhr"
```

## Compare the two ciphers

A Caesar cipher has only 26 possible shifts, so someone can try them all quickly. Vigenere uses a sequence of shifts, but repeated-key patterns still reveal information. Both are reversible because decoding applies the opposite shift.

```python
secret = caesar_encode("Algorithms are fun!", 7)
print("Ciphertext:", secret)
print("\nTrying every Caesar key:")
for possible_shift in range(26):
    print(f"{possible_shift:2}: {caesar_decode(secret, possible_shift)}")
```
